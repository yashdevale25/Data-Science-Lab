To Use the NumPy, Scipy, Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
from scipy import stats
from scipy.stats import norm
```

Step 1: User Input (Common for all libraries)

```python
data = list(map(float, input("Enter numbers separated by space: ").split()))

print("Dataset entered by user:", data)
```

#Statistical Analysis

Program 1: Using NumPy amd SciPy Calculating Mean, Median, Mode, Variance & Standard Deviation

```python
# Convert to NumPy array
data_np = np.array(data)

# Calculations
mean = np.mean(data_np)
median = np.median(data_np)
mode = stats.mode(data_np, keepdims=True)[0][0]
variance = np.var(data_np)
std_dev = np.std(data_np)

# Display results
print("\nStatistical Measures:")
print("Mean (μ):", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance (σ²):", variance)
print("Standard Deviation (σ):", std_dev)
```

#Data Visualization

Program 3: Using Matplotlib (Visualization)

```python
# Plot histogram with better visibility
plt.figure(figsize=(6, 5))          # Increase figure size
plt.hist(data_np, bins=11

, edgecolor='black')

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Dataset Distribution")

plt.grid(True)                       # Add grid for clarity
plt.show()
```

Normal Distribution Graph

```python

x = np.linspace(mean - 3*std_dev, mean + 3*std_dev, 100)

# Calculate normal distribution (PDF)
y = norm.pdf(x, mean, std_dev)

# Plot only the normal curve
plt.figure(figsize=(8,5))
plt.plot(x, y)

# Labels
plt.title("Normal Distribution Curve")
plt.xlabel("Values")
plt.ylabel("Probability Density")

plt.show()
```

#Probabilistic Simulations

Program 4: Coin Toss Probability Simulation

```python
# Number of coin tosses
tosses = int(input("Enter number of coin tosses: "))

# Simulate coin toss (0 = Head, 1 = Tail)
results = np.random.randint(0, 2, tosses)

# Count outcomes
heads = np.sum(results == 0)
tails = np.sum(results == 1)

# Calculate probabilities
print("\nCoin Toss Simulation Results:")
print("Heads:", heads)
print("Tails:", tails)
print("Probability of Head:", heads / tosses)
print("Probability of Tail:", tails / tosses)
```
