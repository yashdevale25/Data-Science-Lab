#Linear Algebra : Without NumPy Library(Manually)

a) Create a 3x3 square matrix and compute its determinant.

```python
# --- Determinant of a 3x3 Matrix ---
A = [[1, 2, 3],
     [0, 1, 4],
     [5, 6, 0]]

def determinant(m):
    a, b, c = m[0]
    d, e, f = m[1]
    g, h, i = m[2]

    det = (a * (e * i - f * h)) - (b * (d * i - f * g)) + (c * (d * h - e * g))
    return det

det_A = determinant(A)
print(f"Determinant of A: {det_A}")
```

b) Find the inverse of the matrix

```python
# --- Inverse of a 3x3 Matrix ---
def inverse(m):
    det = determinant(m)
    if det == 0:
        return "Matrix is singular, no inverse."

    # Matrix of Cofactors
    # We find the determinant of the 2x2 minor for each element and apply sign
    def det2x2(a, b, c, d):
        return a * d - b * c

    # Calculating Adjugate Matrix (Transpose of Cofactors)
    inv = [
        [det2x2(m[1][1], m[1][2], m[2][1], m[2][2]) / det, -(det2x2(m[0][1], m[0][2], m[2][1], m[2][2])) / det, det2x2(m[0][1], m[0][2], m[1][1], m[1][2]) / det],
        [-(det2x2(m[1][0], m[1][2], m[2][0], m[2][2])) / det, det2x2(m[0][0], m[0][2], m[2][0], m[2][2]) / det, -(det2x2(m[0][0], m[0][2], m[1][0], m[1][2])) / det],
        [det2x2(m[1][0], m[1][1], m[2][0], m[2][1]) / det, -(det2x2(m[0][0], m[0][1], m[2][0], m[2][1])) / det, det2x2(m[0][0], m[0][1], m[1][0], m[1][1]) / det]
    ]
    return inv

if det_A != 0:
    print("\nInverse of Matrix A:")
    for row in inverse(A):
        # Using string formatting to truncate to 2 decimals as requested earlier
        print([f"{x:.2f}" for x in row])
else:
    print("\nMatrix A is singular.")
```

c) Solve the following system of linear equations using NumPy: 2x+3y=10, 4x+5y=20

```python
# --- Solving 2x2 System (Cramer's Rule) ---
# 2x + 3y = 10
# 4x + 5y = 20
def solve_2x2(coeffs, consts):
    a, b = coeffs[0]
    c, d = coeffs[1]
    e, f = consts

    D = (a * d) - (b * c)
    if D == 0:
        return "No unique solution"

    Dx = (e * d) - (b * f)
    Dy = (a * f) - (e * c)

    return [Dx / D, Dy / D]

B = [[2, 3], [4, 5]]
C = [10, 20]
sol = solve_2x2(B, C)
print(f"\nSolution (x, y): {sol}")
```

#Linear Algebra : Using NumPy Library

```python
import numpy as np
```

a) Create a 3x3 square matrix and compute its determinant.

```python
import numpy as np

A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])

det = np.linalg.det(A)


deter = int(det * 100) / 100

print("Matrix A:")
print(A)
print(f"Determinant of the matrix: {deter:.2f}")
```

b) Find the inverse of the matrix

```python
# Create a 3x3 matrix
A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])

print("Matrix A:")
print(A)

# Find inverse
if np.linalg.det(A) != 0:
    inverse = np.linalg.inv(A)
    print("Inverse of the matrix:")
    print(inverse)
else:
    print("Inverse does not exist (determinant is zero).")
```

c) Solve the following system of linear equations using NumPy: 2x+3y=10, 4x+5y=20

```python
# Coefficient matrix
A = np.array([[2, 3],
              [4, 5]])

# Constant matrix
B = np.array([10, 20])

# Solve the equations
solution = np.linalg.solve(A, B)

print("Solution of the given system:")
print("x =", solution[0])
print("y =", solution[1])
```
